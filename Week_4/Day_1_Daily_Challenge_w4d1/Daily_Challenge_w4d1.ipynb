2. Data Visualization with Matplotlib:
Create an interactive line chart to show sales trends over the years.
Build an interactive map to visualize sales distribution by country.

```python
import matplotlib.pyplot as plt
import pandas as pd

superstore_dataset = pd.read_excel("/home/daniella/code/DataAnalytics-2026/Week_4/Day_1_Exercise_XP_w4d1/US_Superstore_data.xls")

import plotly.express as px


superstore_dataset['Order Date'] = pd.to_datetime(superstore_dataset['Order Date'])
superstore_dataset['Year'] = superstore_dataset['Order Date'].dt.year


sales_by_year = superstore_dataset.groupby(['Year', 'Category'], as_index=False)['Sales'].sum()

fig = px.line(
    sales_by_year,
    x='Year',
    y='Sales',
    color='Category',          # Adds interactive legend items to toggle categories on/off
    markers=True,              # Shows data points on the line
    title='Superstore Sales Trends Over the Years by Category',
    labels={'Sales': 'Total Sales ($)', 'Year': 'Order Year'},
    template='plotly_white'
)

# # 5. Format the X-axis to show clean integer years and refine hover tooltips
fig.update_xaxes(dtick=1)
fig.update_traces(
    hovertemplate='<b>Year:</b> %{x}<br><b>Total Sales:</b> $%{y:,.2f}<extra></extra>'
)

# 6. Display the interactive chart
fig.show()

```

3. Data Visualization with Seaborn:
Use Seaborn to generate a bar chart showing top 10 products by sales.
Create a scatter plot to analyze the relationship between profit and discount.

```python
import seaborn as sns

top_10_products = (
    superstore_dataset.groupby('Product Name', as_index=False)['Sales']
    .sum()
    .sort_values(by='Sales', ascending=False)
    .head(10)
)

sns.set_theme(style='whitegrid')

bar_chart = sns.catplot(
    data=top_10_products,
    x='Sales',
    y='Product Name',
    hue='Product Name',
    kind='bar',
    palette='viridis',
    legend=False,
    height=6,
    aspect=1.8
)

# 3. Set title and axis labels using Seaborn's .set() method
bar_chart.set(
    title='Top 10 Products by Total Sales',
    xlabel='Total Sales ($)',
    ylabel='Product Name'
);
```
