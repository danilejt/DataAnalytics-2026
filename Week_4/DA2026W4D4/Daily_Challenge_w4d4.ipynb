Use NumPy to generate a synthetic dataset representing average monthly temperatures (in degrees Celsius) for 12 months across 10 different cities. The temperatures should range from -5 to 35 degrees.

Convert this NumPy array into a Pandas DataFrame, adding city names as index and months as columns.
2. Data Analysis:

Hint 1: Calculate the annual average temperature using DataFrame.mean(axis).
Hint 2: Find the city with the highest and lowest average temperature using idxmax() and idxmin() methods.

Calculate the annual average temperature for each city.

Identify the city with the highest and lowest average temperature for the year.
3. Data Visualization:

Deliverables:
A Jupyter Notebook containing all the code for data generation, analysis, and visualization.
A brief report within the notebook summarizing your findings, including the city with the highest and lowest average temperatures and any interesting trends observed in the data.


```python
import numpy as np
import pandas as pd

np.random.seed(42)
temp_data = np.random.uniform(low=-5.0, high=35.0, size=(10, 12))
temp_data = np.round(temp_data, 1)
cities = [
    "Jerusalem", "New York", "London", "Paris", "Sydney",
    "Toronto", "Berlin", "Madrid", "Seoul", "Cairo"
]

months = [
    "Jan", "Feb", "Mar", "Apr", "May", "Jun",
    "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"
]


df_temperatures = pd.DataFrame(data=temp_data, index=cities, columns=months)

print(df_temperatures)
```

```python
# Hint 1: Calculate the annual average temperature using DataFrame.mean(axis).

df_temperatures_mean_month = df_temperatures.mean(axis = 0)
print(df_temperatures_mean_month)
```

```python
# Hint 2: Find the city with the highest and lowest average temperature using idxmax() and idxmin() methods.

df_temperatures_mean_city = df_temperatures.mean(axis = 1)
highest_city = df_temperatures_mean_city.idxmax()
lowest_city = df_temperatures_mean_city.idxmin()

print(f"Highest average temperature: {highest_city}")
print(f"Lowest average temperature: {lowest_city}")
```

```python
import matplotlib.pyplot as plt

df_transposed = df_temperatures.transpose()
ax = df_temperatures.transpose().plot(kind="line", linestyle = ":", marker="^", figsize=(10, 5))


plt.title("Monthly Average Temperature by City")
plt.xlabel("Month")
plt.ylabel("Temperature (°C)")
plt.xticks(
    ticks=range(len(df_transposed.index)), labels=df_transposed.index)

plt.grid(True, linestyle="--", alpha=0.6)
plt.legend(title="City", bbox_to_anchor=(-0.1, 1), loc="upper right")

# Display the chart
plt.show()
```

```python
# Calculate key summary metrics
highest_temp = df_temperatures_mean_city.max()
lowest_temp = df_temperatures_mean_city.min()
temp_spread = highest_temp - lowest_temp

# Calculate each city's seasonal range (Max month - Min month)
seasonal_range = df_temperatures.max(axis=1) - df_temperatures.min(axis=1)
most_variable_city = seasonal_range.idxmax()
most_stable_city = seasonal_range.idxmin()

print("=== ANNUAL TEMPERATURE REPORT ===")
print(
    f"• Highest Average Temperature: {highest_city} ({highest_temp:.2f}°)"
)
print(f"• Lowest Average Temperature:  {lowest_city} ({lowest_temp:.2f}°)")
print(f"• Annual Average Spread:       {temp_spread:.2f}°")
print(
    f"• Largest Seasonal Swing:      {most_variable_city}"
    f" ({seasonal_range[most_variable_city]:.2f}° range)"
)
print(
    f"• Most Stable Climate:         {most_stable_city}"
    f" ({seasonal_range[most_stable_city]:.2f}° range)"
)
```
