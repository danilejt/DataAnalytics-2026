🌟 Exercise 2 : Type Conversion and Array Operations
Instructions
Convert a list [3.14, 2.17, 0, 1, 2] into a NumPy array and convert its data type to integer.

Expected Output:



array([3, 2, 0, 1, 2])

```python
import numpy as np

my_list = [3.14, 2.17, 0, 1, 2]

array2 = np.array(my_list).astype(int)

print(array2)

```

🌟 Exercise 6 : Reversing elements
Instructions
Reverse the order of elements in a given 1D NumPy array (first element becomes last).

Expected Output:

array([9, 8, 7, 6, 5, 4, 3, 2, 1, 0])

```python
import numpy as np
array6 = np.arange(0, 10)
array6_reversed = array6[::-1]
print(array6_reversed)
```


🌟 Exercise 8 : Simple Aggregate Funcs
Instructions
Find the sum and average of a given 1D array.

Expected Output:

```python
import numpy as np
array8 = np.arange(1, 6)
sum = np.sum(array8)
mean = np.mean(array8)
print(sum)
print(mean)
```

🌟 Exercise 10 : Conditional Selection of Values
Instructions
Extract all odd numbers from a given NumPy array.

Expected Output:



array([1, 3, 5, 7, 9])

```python
import numpy as np

array10 = np.random.randint(1, 101, size=10)

odd_numbers = array10[array10 % 2 != 0]

print(odd_numbers)

```
