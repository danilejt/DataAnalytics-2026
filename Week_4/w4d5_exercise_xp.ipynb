🌟 Exercise 1 : Matrix Operations
Instructions
In this exercise, you’ll work with a 3x3 matrix. Here’s a brief explanation of the concepts:

Determinant: The determinant is a value that can be computed from the elements of a square matrix. It provides important information about the matrix, such as whether it has an inverse, and is used in various areas like linear algebra and calculus. To understand more about it you can watch this video.
Inverse of a Matrix: The inverse of a matrix is a matrix that, when multiplied with the original matrix, results in an identity matrix. Not all matrices have inverses. The inverse is crucial in solving systems of linear equations.
Create a 3x3 matrix and perform the following operations:

Calculate the determinant.
Find the inverse of the matrix.

```python
import pandas as pd
import numpy as np

M = np.random.randint(1, 10, size=(3, 3)).astype(int)
print("Random 3x3 Matrix:\n", M)

# Unpack elements for readability
a, b, c = M[0]
d, e, f = M[1]
g, h, i = M[2]

# 2. Calculate the determinant manually
det = a * (e * i - f * h) - b * (d * i - f * g) + c * (d * h - e * g)
print("\nDeterminant:", det)

# 3. Find the inverse using the adjugate matrix (if determinant is not 0)
if det != 0:
    # Compute the 9 cofactors
    c11 =  (e * i - f * h)
    c12 = -(d * i - f * g)
    c13 =  (d * h - e * g)

    c21 = -(b * i - c * h)
    c22 =  (a * i - c * g)
    c23 = -(a * h - b * g)

    c31 =  (b * f - c * e)
    c32 = -(a * f - c * d)
    c33 =  (a * e - b * d)

    # Transpose the cofactors into the adjugate matrix and divide by determinant
    adjugate = np.array([
        [c11, c21, c31],
        [c12, c22, c32],
        [c13, c23, c33]
    ])
    inverse = adjugate / det

    print("\nInverse Matrix:\n", np.round(inverse, 3))

    print("\nVerification (M @ Inverse):\n", np.round(M @ inverse, 2))
else:
    print("\nDeterminant is 0; this random matrix has no inverse. Run again!")
```

🌟 Exercise 6 : Basic Hypothesis Testing
Instructions
Create a sample dataset to test the effectiveness of a new training program on employee productivity:

Given a dataset, formulate a simple hypothesis and test it using basic statistical functions in NumPy.

```python
import numpy as np

# Productivity scores of employees before the training program
productivity_before = np.random.normal(loc=50, scale=10, size=30)

# Productivity scores of the same employees after the training program
productivity_after = productivity_before + np.random.normal(loc=5, scale=3, size=30)

# Your task is to formulate a hypothesis regarding the training program's effectiveness
# and test it using basic statistical functions in NumPy.
diff = productivity_after - productivity_before
print(diff)
print(np.mean(diff))
print(np.std(diff))

t_stat = np.mean(diff) / (np.std(diff, ddof=1) / np.sqrt(len(diff)))

# Calculate one-tailed p-value using normal approximation
p_value = 0.5 * math.erfc(t_stat / math.sqrt(2))

print(f't-statistic: {t_stat:.2f}')
print(f'p-value:     {p_value:.4e}')

alpha = 0.05

if p_value < alpha:
  print('Decision: Reject the Null Hypothesis (H0).')
  print('Conclusion: The training program had a statistically significant positive effect.'
  )
else:
  print('Decision: Fail to reject the Null Hypothesis (H0).')
  print('Conclusion: There is not enough evidence that the training was effective.')
```

🌟 Exercise 10 : Basic Visualization
Instructions
Use Matplotlib to visualize a simple dataset created with NumPy (e.g., a line graph of random numbers).

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import time
import random
import math


np.random.seed(42)               # Ensures the random numbers are reproducible
x = np.arange(1, 11)             # 50 points on the x-axis (1 to 50)
y = np.random.randn(10).cumsum() # Cumulative sum of 50 random numbers (a random walk)

# 2. Create the line plot
plt.figure(figsize=(8, 4))
plt.plot(x, y, marker='^', linestyle='--', color='blue', markersize=4)

# 3. Add labels and styling
plt.title("Random Dataset (NumPy + Matplotlib)")
plt.xlabel("Step")
plt.ylabel("Value")
plt.grid(True, linestyle='--', alpha=0.6)

# 4. Display the chart
plt.tight_layout()
plt.show()
```
