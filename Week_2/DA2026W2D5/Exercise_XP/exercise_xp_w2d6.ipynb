Data analysis is the systematic process of inspecting, cleaning, transforming, and modeling raw data to discover useful information, draw conclusions, and support decision-making. It bridges the gap between raw data collection and actionable insight.
Data analysis is divided into two primary categories: (1) Quantitative Analysis: Examining numerical or measurable data, but also (2) 
Qualitative Analysis: Examining non-numerical data. As a math teacher myself currently, I obviously graviate towards the quantitative. 

Data analysis transforms fragmented, unorganized figures into strategic assets. Without it, organizations and researchers are left with noise rather than knowledge. Data analysis helps with informed decision-making, by replacing intuition and guesswork with empirical evidence, reducing risk when committing financial or operational resources. It also, depending on the nature of the analysis and the company etc, helps with operational efficiency. How? Identifies bottlenecks, redundant processes, and cost-saving opportunities within workflows or supply chains. 

From my research, a standard data analysis workflow generally follows these sequential phases: 1. Define the Objective: Clearly state the problem, question, or hypothesis to be answered. 2. Data Collection: Gather quantitative or qualitative data from primary sources (surveys, experiments, interviews) or secondary sources (databases, public logs). 3. Data Cleaning: Inspect the data to fix errors, handle missing values, remove duplicates, and standardize formats.4. Data Exploration & Analysis: Apply statistical tools, sorting, coding, or modeling frameworks to uncover trends and relationships. 5. Interpretation & Visualization: Translate findings into all types of visual chart

Objective: Learn to identify different data types.
Task:
For the datasets from the previous exercise, categorize each column of it as either quantitative or qualitative and explain your reasoning.

```python
datasets = {"Credit Card": df_credit_card_approval, "Mental Health": df_global_trends_in_mental_health, "Sleep": df_how_much_sleep}

for name, data in datasets.items():
    print(f" ***** {name} *****")
    for column in data.columns:
        if not pd.api.types.is_numeric_dtype(data[column]) or data[column].nunique() <= 2 or "zip" in column.lower():
            category = "Qualitative"
        else:
            category = "Quantitative"
        print(f"  {column}: {category} (dtype: {data[column].dtype}, unique: {data[column].nunique()})")
    print()
```

🌟Exercise 5: Basic Observation Skills in Data Analysis


Objective: Develop observation skills for data analysis.
Task:
Load the How Much Sleep Do Americans Really Get? dataset into Jupyter or Google Colab.
Identify columns that could be interesting for a specific type of analysis (e.g., trend analysis, group comparison) and explain your choice.
Tools: Jupyter Notebook, Python with Pandas library.


```python
print(df_how_much_sleep.info())
#columns that will be interesting: comparing men v. women and how much sleep they get, comparing how sleep has changed by the year, or comparing the type of days and eeing how sleep changes by day

# 2. Classify columns (Quantitative vs. Qualitative)
print("\n***  COLUMN CLASSIFICATION ***")
for column in df_how_much_sleep.columns:
    if (not pd.api.types.is_numeric_dtype(df_how_much_sleep[column])
        or df_how_much_sleep[column].nunique() <= 2
        or column.lower() in ["id", "index", "year"]):
        category = "Qualitative"
    else:
        category = "Quantitative"
    print(f" {column}: {category} (dtype: {df_how_much_sleep[column].dtype}, unique: {df_how_much_sleep[column].nunique()})")

# 3. Trend Analysis: Average Sleep Over Time (by Year)
print("\n***  TREND ANALYSIS: AVG SLEEP BY YEAR ***")
trend_analysis = df_how_much_sleep.groupby("Year")["Avg hrs per day sleeping"].mean().reset_index()
display(trend_analysis)

# 4. Group Comparison: By Age Group and Sex
print("\n*** 4. GROUP COMPARISON: BY AGE GROUP & SEX ***")
demographic_comparison = (
    df_how_much_sleep.groupby(["Age Group", "Sex"])["Avg hrs per day sleeping"].mean()
   .unstack()  # Makes Men and Women side-by-side columns for easy reading
   .reset_index()
)
display(demographic_comparison)

# 5. Lifestyle Comparison: Weekdays vs. Weekends/Holidays
print("\n*** SCHEDULE COMPARISON: BY TYPE OF DAYS ***")
day_type_comparison = (
    df_how_much_sleep.groupby("Type of Days")["Avg hrs per day sleeping"]
    .agg(["mean", "min", "max"])
    .reset_index()
)
display(day_type_comparison)
```
