```python
from sklearn import preprocessing
from sklearn.preprocessing import MinMaxScaler


scaler = MinMaxScaler()

data_science_salaries_csv['salary'] = scaler.(data_science_salaries_csv[['salary']])


data_science_salaries_csv[['salary']].head()
```

Implement dimensionality reduction like Principal Component Analysis (PCA) or t-SNE to reduce the number of features (columns) in the dataset.

```python
from sklearn.decomposition import PCA

numeric_df = data_science_salaries_csv.select_dtypes(include=['number'])

numeric_df = numeric_df.fillna(numeric_df.median())

pca = PCA(n_components=2)

pca_results = pca.fit_transform(numeric_df)

pca_df = pd.DataFrame(
    data=pca_results, 
    columns=['', 'salary']
)

# 6. View the new, compressed dataset
print(f"Original shape: {numeric_df.shape} -> New shape: {pca_df.shape}")
pca_df.head()
```

Group the dataset by the ‘experience_level’ column and calculate the average and median salary for each experience level (e.g., Junior, Mid-level, Senior).

```python
# Group by experience level and calculate mean (average) and median of the salary column
salary_summary = data_science_salaries_csv.groupby('experience_level')['salary'].agg(['mean', 'median'])

# Rename columns for clarity
salary_summary.columns = ['Average Salary', 'Median Salary']

# Display the summary table
salary_summary
```
